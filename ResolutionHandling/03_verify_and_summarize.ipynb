```python
import numpy as np
import pandas as pd
from pathlib import Path
```

```python
BASE = Path('/home/msp25gd/ResearchProjectMSc/ResolutionHandling')
TARGET = BASE / 'processed_candidates'
LOG = BASE / 'logs'
mixed_groups = np.load(BASE / 'mixed_groups.npy', allow_pickle=True).tolist()

ok = 0
bad = []

for g in mixed_groups:
    meta_path = TARGET / str(g) / 'meta' / 'group_df.pkl'
    if not meta_path.exists():
        bad.append((str(g), 'missing_meta'))
        continue

    df = pd.read_pickle(meta_path)
    nuniq = pd.to_numeric(df['SPEC_RES'], errors='coerce').nunique(dropna=True)
    if nuniq == 1:
        ok += 1
    else:
        bad.append((str(g), f'nunique={nuniq}'))

print(f'Mixed groups expected: {len(mixed_groups)}')
print(f'Mixed groups now uniform: {ok}')
print(f'Mixed groups with issues: {len(bad)}')

if bad:
    print('\nFirst issues:')
    for item in bad[:10]:
        print(' -', item[0], item[1])

status_log = pd.read_csv(LOG / 'processed_groups_log.csv')
print('\nStatus counts:')
print(status_log['status'].value_counts().to_string())
```
