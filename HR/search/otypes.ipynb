```python
import pandas as pd
from astroquery.simbad import Simbad

def get_otypes(star):

    # Custom Simbad search (field names updated for new TAP-based SIMBAD API)
    custom_simbad = Simbad()
    custom_simbad.TIMEOUT = 10
    custom_simbad.add_votable_fields('otypes', 'rvz_radvel', 'otype', 'plx_value', 'sp_type')

    sim = custom_simbad.query_objects([star])
    try:
        df = sim.to_pandas()
        # Keep only the 6 columns that match the 6 colWidths in the table display
        keep = ['main_id', 'otypes.otype_txt', 'otype', 'sp_type', 'rvz_radvel', 'plx_value']
        avail = [c for c in keep if c in df.columns]
        return df[avail]
    except Exception as e:
        print('SIMBAD error for "{}": {}'.format(star, e))
        return



# star = 'beta pic'
# print(get_otypes(star))
```
